c. What is the role of the `top20` index in the graph? What would have happened if we had not updated the index?

The top 20 index is especially important for graphing because it means that we are only plotting the 20 largest earthquakes.  If we had called ".plot.bar" on "eqk" (the full earthquake dataframe) instead of "top20" (the 20 largest earthquakes), python would attempt to create a barplot of every single earthquake in the database (of which there are 118,398).  If this produced a graph at all, it would be completely unreadable.  Therefore the top20 index allows to create a graph that is actually useful and readable.
